## Boolean arrays - A way to access a given set of elements in an array

This section provides some useful background that you may choose employ for exercise #6.2 below.

If you have an array, you can perform queries based on values of that array.  These queries return "Boolean arrays", which will contain True or False values depending on whether the element of the array meets the criteria.  You can then use these Boolean arrays to only access the value you would

```python
#this example shows you how to create and use a boolean array with a single argument
a = np.array([1,2,3,4,5])           #create an array
#perform a query.
#I always end the output of a Boolean query out with "flag" so that I know it's a 
#Boolean array
asmallflag = a < 4           

#you can see that only the first three elements of asmall are "True"
print('A boolean array that selects small a values:', asmallflag)

#now using 'abig' as an array of indices for 'a' you can print out only the elements of a 
#that meet your query, that is, those that have True indices
print('Only the elemens of a for which the query yielded a True:', a[asmallflag])

#Now let's make a second array with the same length as a
b = np.array(['aa','bb','cc','dd','ee'])

# in this case, b could be names of object in a table and a could be some value associated 
#with those objects.  Since they have the same length, same order, **and** refer to the same set of data
#we can apply the 'abig' Boolean array to find out which elements of b are selected when we 
#perform our query on a
print('Only the elements of b that were selected by the previous query:', b[asmallflag])

#that's pretty cool!  But wait, there's more
#you can also make Boolean arrays by combining different conditions.  You can do this two different
#wasy.  
#First, you can set up a single query with two conditions, each enclosed in ()
amidflag = (a>2) & (a<4)
print('The elements of a and b that meet the new criteria: ', a[amidflag], b[amidflag])

#but I can also make individual Boolean arrays and combine them with a logical AND (&) command.
#Only elements that are True in both input arrays are true in the output array
abigflag = a>2
amidflag2 = abigflag & asmallflag
print('The results of combining the two previous selections:',amidflag2, a[amidflag2])

#and finally, we can assign only the values that satisfy a boolean command to a new array.
#in this case, only values of a with True indicies in amidflag2 will be assigned to onlymidvalues
onlymidvalues = a[amidflag2]
print(onlymidvalues)

#in the exercise below, we will use this technique to print out the flux at a point
#on the spectrum with a given wavelength
```

### TM
the g-r of star1 is more negative than that of star3, meaning star1 is bluer and star3 is redder. This is an extremely rudimentary way of calculating this, as it only factors in the flux density at the central wavelength, even though the bands are quite broad.
My methodology for this next part is to first get the wavelength bounds of the bands, defined by the full width at half maximum FWHM. Then I'll calculate the median from all of my flam values in each band (not mean, as there's quite a lot of outliers from absorption or error) and, after conversion, use those as my fnu values. 
I will be ignoring the fact that the bands aren't really symmetrical. If I were to incorporate that, I would forgo my FWHM method and, in my calculations for the average fnu of the band, I would weight each fnu value by the band "strength" at that wavelength. Since the band "strength" values are effectively the fraction of light let in at that wavelength, I think it would be possible to just multiply the fnu values by the band "strength" value nearest to that wavelength.

```python
#### Write code here for 6.2


# finding FWHM for g and r bands

# half of the maximum "flam" value (not actually flam, it's a % value of flux let through filter)
gHM = np.max(gband.response/2)
rHM = np.max(rband.response/2)

# same method as used previously to find closest lam value
gHM_vertdiff = abs(gband.response - gHM)
rHM_vertdiff = abs(rband.response - rHM)

# this is just so I can get the values closest to the HM
# on both sides of each band
gband_lo_flag = gband.wavelength < 4500
gband_hi_flag = gband.wavelength > 4500
rband_lo_flag = rband.wavelength < 6000
rband_hi_flag = rband.wavelength > 6000

# getting the index array for the FWHM bounds for both bands
gFWHMlo_flam = np.argmin(gHM_vertdiff[gband_lo_flag])
gFWHMhi_flam = np.argmin(gHM_vertdiff[gband_hi_flag])
rFWHMlo_flam = np.argmin(rHM_vertdiff[rband_lo_flag])
rFWHMhi_flam = np.argmin(rHM_vertdiff[rband_hi_flag])

# the boundary values for the FWHM of either band
gFWHMlo_lam = gband.wavelength[gband_lo_flag][gFWHMlo_flam]
gFWHMhi_lam = gband.wavelength[gband_hi_flag][gFWHMhi_flam]
rFWHMlo_lam = rband.wavelength[rband_lo_flag][rFWHMlo_flam]
rFWHMhi_lam = rband.wavelength[rband_hi_flag][rFWHMhi_flam]


# the last of the three conditions is so that it's less likely for the median to end up
# as one of the outlier points, as it forces them to be excluded
gFWHM_s1_flag = (star1['lambda'] > gFWHMlo_lam) & (star1['lambda'] < gFWHMhi_lam) & (star3['flam'] > 2*10**-9)
rFWHM_s1_flag = (star1['lambda'] > rFWHMlo_lam) & (star1['lambda'] < rFWHMhi_lam) & (star3['flam'] > 1*10**-9)
gFWHM_s3_flag = (star3['lambda'] > gFWHMlo_lam) & (star3['lambda'] < gFWHMhi_lam) & (star3['flam'] > 2.75*10**-9)
rFWHM_s3_flag = (star3['lambda'] > rFWHMlo_lam) & (star3['lambda'] < rFWHMhi_lam) & (star3['flam'] > 3*10**-9)

gflam_star1 = np.median(star1['flam'][gFWHM_s1_flag])
rflam_star1 = np.median(star1['flam'][rFWHM_s1_flag])
gflam_star3 = np.median(star3['flam'][gFWHM_s3_flag])
rflam_star3 = np.median(star3['flam'][rFWHM_s3_flag])

# for the wavelength value corresponding to the flam values, I'll just use
# the central wavelength value of the FWHM
glam = (gFWHMlo_lam+gFWHMhi_lam)/2
rlam = (rFWHMlo_lam+rFWHMhi_lam)/2

print(gflam_star1,rflam_star1,gflam_star3,rflam_star3)
print(glam,rlam)


#I'm assuming that you did your calculation for fnu above this line
fnu_star1_gband = gflam_star1 * (glam**2/(3*10**18))
fnu_star3_gband = gflam_star3 * (glam**2/(3*10**18))

fnu_star1_rband = rflam_star1 * (rlam**2/(3*10**18))
fnu_star3_rband = rflam_star3 * (rlam**2/(3*10**18))


#print your fnu's here with the following format
print(f'fnu of star1 at the g-band = {fnu_star1_gband:5.2e} erg/s/cm^2/Hz')
print(f'fnu of star1 at the r-band = {fnu_star1_rband:5.2e} erg/s/cm^2/Hz')
print(f'fnu of star3 at the g-band = {fnu_star3_gband:5.2e} erg/s/cm^2/Hz')
print(f'fnu of star3 at the r-band = {fnu_star3_rband:5.2e} erg/s/cm^2/Hz')

#AB mag for each filter and star
#your calculations

ABmag_star1_gband = -2.5 * np.log10(fnu_star1_gband) - 48.6
ABmag_star1_rband = -2.5 * np.log10(fnu_star1_rband) - 48.6
ABmag_star3_gband = -2.5 * np.log10(fnu_star3_gband) - 48.6
ABmag_star3_rband = -2.5 * np.log10(fnu_star3_rband) - 48.6

#print your AB mags here with the following format
print(f'AB mag of star1 at the g-band = {ABmag_star1_gband:4.2f}')
print(f'AB mag of star1 at the r-band = {ABmag_star1_rband:4.2f}')
print(f'AB mag of star3 at the g-band = {ABmag_star3_gband:4.2f}')
print(f'AB mag of star3 at the r-band = {ABmag_star3_rband:4.2f}')

#g-r color for each star
#your calculations

gr_star1 = ABmag_star1_gband - ABmag_star1_rband
gr_star3 = ABmag_star3_gband - ABmag_star3_rband

#print your colors here with the following format
print(f'g-r color of star1 = {gr_star1:4.2f}')
print(f'g-r color of star3 = {gr_star3:4.2f}')
```
